```python
# 1. Variables and Dynamic Typing
a = 2
b = a * a
name = "jo"
print(f"a = {a}, b = {b}, name = {name}")

counter = 100
miles = 1000.0
count = counter * 2
travel = miles / 2.5
print(f"count: {count}, travel: {travel}")

# 2. Lists and Mutation
z = [1, 2, 3]
z.extend([4, 5, 6])
z.append(0)
print(f"Extended & appended list z: {z}")

integer_list = [1, 2, 3, 4, 5, 6, 7, 8, 9]
print(f"Index 2 element: {integer_list[2]}")
print(f"Is 23 in integer_list? {23 in integer_list}")

# 3. Slicing
first_two = integer_list[:2]
four_to_end = integer_list[4:]
print(f"First two items: {first_two}")
print(f"Four to end: {four_to_end}")
```

```python
from collections import Counter

# 1. List Comprehensions (Filtering even and odd numbers)
x = [1, 2, 3, 4, 5]
evens = [i for i in range(5) if i % 2 == 0]
odds = [i for i in range(5) if i % 2 > 0]
print(f"Evens: {evens}")
print(f"Odds: {odds}")

# 2. Tuples (Immutable sequences)
mytuple1 = (1, 2)
mytuple2 = 3, 4
print(f"Tuple 1: {mytuple1}, Tuple 2: {mytuple2}")

# 3. Dictionaries & Counters
grades = {"jo": 80, "tim": 90}
print(f"Jo's Grade: {grades['jo']}")

c = Counter([0, 1, 2, 0])
print(f"Counter result: {c}")

# 4. Sets (Fast membership testing)
stopwords = {"a", "an", "the", "yet", "you"}
print(f"'you' in stopwords: {'you' in stopwords}")
print(f"'gong' in stopwords: {'gong' in stopwords}")
```

```python
import random
import re

# 1. Functions & Lambdas
def mysquare(x):
    return x * x

add_five = lambda number: number + 5

print(f"mysquare(3) = {mysquare(3)}")
print(f"add_five(4) = {add_five(4)}")

# 2. Random Seed & Regular Expressions
random.seed(10)
print(f"Random value: {random.random():.4f}")

sample_str = "an example word:cat !!"
match = re.search(r"word:\w\w\w", sample_str)
if match:
    print(f"Regex found: {match.group()}")

# 3. Object-Oriented Class Implementation
class Employee:
    empCount = 0

    def __init__(self, name: str, salary: float):
        self.name = name
        self.salary = salary
        Employee.empCount += 1

    def display_employee(self):
        print(f"Name: {self.name}, Salary: ${self.salary:,.2f}")

emp1 = Employee("Zara", 2000)
emp2 = Employee("Manni", 5000)
emp1.display_employee()
emp2.display_employee()
print(f"Total Employees: {Employee.empCount}")
```

```python
def fahrenheit_to_centigrade(f_temp: float) -> float:
    """Converts Fahrenheit temperature to Centigrade."""
    return (f_temp - 32) / 1.8

def centigrade_to_fahrenheit(c_temp: float) -> float:
    """Converts Centigrade temperature to Fahrenheit."""
    return (c_temp * 1.8) + 32

# Test Suite Demonstrating Correctness
test_f = [32.0, 68.0, 98.6, 212.0]
test_c = [0.0, 20.0, 37.0, 100.0]

print("=== Exercise 1: Fahrenheit to Centigrade ===")
for f in test_f:
    c = fahrenheit_to_centigrade(f)
    print(f"{f:6.1f} °F  -->  {c:6.2f} °C")

print("\n=== Exercise 1: Centigrade to Fahrenheit ===")
for c in test_c:
    f = centigrade_to_fahrenheit(c)
    print(f"{c:6.1f} °C  -->  {f:6.2f} °F")
```

```python
def calculate_average(numbers: list) -> float:
    """Calculates arithmetic mean using an explicit for loop."""
    if not numbers:
        return 0.0

    total_sum = 0.0
    count = 0

    for num in numbers:
        total_sum += num
        count += 1

    return total_sum / count

# Test Scenarios
sample1 = [10, 20, 30, 40, 50]
sample2 = [14.5, 22.3, 89.1, 45.0, 67.8]
sample3 = list(range(1, 101))

print("=== Exercise 2: Average Calculation Test Runs ===")
print(f"Sample 1: {sample1}")
print(f"Calculated Average: {calculate_average(sample1):.2f}\n")

print(f"Sample 2: {sample2}")
print(f"Calculated Average: {calculate_average(sample2):.2f}\n")

print(f"Sample 3 (Integers 1 to 100):")
print(f"Calculated Average: {calculate_average(sample3):.2f} (Expected: 50.50)")
```
