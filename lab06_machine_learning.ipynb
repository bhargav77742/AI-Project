```python
import pandas as pd
import matplotlib.pyplot as plt

# Exercise 5: Read dataset (semicolon delimiter)
file_name = 'student-mat.csv'
df = pd.read_csv(file_name, sep=';')

# Exercise 6: Structural dimensions
print("DataFrame Shape (Rows, Columns):", df.shape)
print("\nFeature Columns:\n", df.columns.tolist())

# Exercise 7: Summary statistics
print("\nSummary Statistics:")
display(df.describe())

# Exercise 8 & 9: Data integrity checks
print("\nMissing Values Count:\n", df.isnull().sum().max())
print("Duplicate Rows Count:", df.duplicated().sum())

# Exercise 10: Column datatypes
print("\nData Types:\n", df.dtypes)

# Exercise 11: Correlation Heatmap
numerical_cols = df.select_dtypes(include=['int64', 'float64'])
correlations = numerical_cols.corr()

plt.figure(figsize=(10, 8))
plt.imshow(correlations, cmap='coolwarm', interpolation='nearest')
plt.colorbar()
plt.xticks(range(len(correlations)), correlations.columns, rotation=90)
plt.yticks(range(len(correlations)), correlations.columns)
plt.title('Correlation Heatmap of Student Performance Dataset')
plt.tight_layout()
plt.show()

# Exercise 12 & 13: Distribution inspection
print("\nValue Counts for Sex:\n", df['sex'].value_counts())
print("\nValue Counts for Address:\n", df['address'].value_counts())

plt.figure(figsize=(7, 4.5))
plt.hist(df['age'], bins=8, color='#38BDF8', edgecolor='black', rwidth=0.9)
plt.title('Distribution Histogram of Student Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt

# Exercise 14: Define Feature X and Target y
X = df[['G1']]
y = df['G3']

# 80/20 Train-Test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Fit Ordinary Least Squares model
model_g1 = LinearRegression()
model_g1.fit(X_train, y_train)
y_pred = model_g1.predict(X_test)

# Exercise 15: Evaluation Metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"Mean Squared Error (MSE) : {mse:.4f}")
print(f"R-squared Score (R2)     : {r2:.4f}")
print(f"Learned Intercept        : {model_g1.intercept_:.4f}")
print(f"Learned Slope (G1)       : {model_g1.coef_[0]:.4f}")

# Exercise 16: Scatter plot with fitted regression line
plt.figure(figsize=(7, 4.5))
plt.scatter(X_test, y_test, color='#0F172A', alpha=0.75, label='Actual Observations')
plt.plot(X_test, y_pred, color='#EF4444', linewidth=2.5, label=f'Regression Line (R² = {r2:.2f})')
plt.title('Single Variable Linear Regression: G1 vs. Final Grade G3')
plt.xlabel('First Period Grade (G1)')
plt.ylabel('Final Grade (G3)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Exercise 18: Multivariable Linear Regression
multi_features = ['G1', 'G2', 'age', 'traveltime', 'studytime', 'failures']
X_multi = df[multi_features]

X_tr_m, X_te_m, y_tr_m, y_te_m = train_test_split(X_multi, y, test_size=0.2, random_state=0)

multi_model = LinearRegression()
multi_model.fit(X_tr_m, y_tr_m)
y_pred_multi = multi_model.predict(X_te_m)

print("=== EXERCISE 18: MULTIVARIABLE REGRESSION METRICS ===")
print(f"Multivariable MAE : {mean_absolute_error(y_te_m, y_pred_multi):.4f}")
print(f"Multivariable MSE : {mean_squared_error(y_te_m, y_pred_multi):.4f}")
print(f"Multivariable R2  : {r2_score(y_te_m, y_pred_multi):.4f}")

# Exercise 19: External Benchmark (California Housing Dataset)
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing(as_frame=True)
df_h = housing.frame
X_h = df_h[['MedInc', 'HouseAge', 'AveRooms', 'AveBedrms', 'Population']]
y_h = df_h['MedHouseVal']

X_h_train, X_h_test, y_h_train, y_h_test = train_test_split(X_h, y_h, test_size=0.2, random_state=42)
h_model = LinearRegression().fit(X_h_train, y_h_train)
h_preds = h_model.predict(X_h_test)

print("\n=== EXERCISE 19: CALIFORNIA HOUSING REGRESSION PIPELINE ===")
print(f"Housing Model MAE: {mean_absolute_error(y_h_test, h_preds):.4f}")
print(f"Housing Model MSE: {mean_squared_error(y_h_test, h_preds):.4f}")
print(f"Housing Model R2 : {r2_score(y_h_test, h_preds):.4f}")
```
