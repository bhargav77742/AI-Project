```python
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

def preprocess_text(input_text):
    """
    NLP Preprocessing Pipeline:
    1. Lowercase normalization (Ex 5)
    2. Punctuation removal (Ex 6)
    3. Word tokenization (Ex 7)
    4. Stop-word filtering (Ex 8)
    """
    # Exercise 5: Lowercase conversion
    lower_text = input_text.lower()
    
    # Exercise 6: Punctuation elimination
    clean_text = lower_text.translate(str.maketrans('', '', string.punctuation))
    
    # Exercise 7: Tokenization
    tokens = word_tokenize(clean_text)
    
    # Exercise 8: Stop-word filtering
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [w for w in tokens if w not in stop_words]
    
    return {
        "original": input_text,
        "clean_text": clean_text,
        "tokens": tokens,
        "filtered": filtered_tokens
    }

# Baseline Text from Exercise 4
base_text = "This is a simple example of text preprocessing. It involves cleaning and organizing raw text data into a format suitable for analysis."
res_base = preprocess_text(base_text)

print("=== BASELINE TEXT PREPROCESSING (EXERCISES 4-8) ===")
print("Original Text:\n", res_base["original"])
print("\nTokens after Tokenization:\n", res_base["tokens"])
print("\nFiltered Tokens (Stopwords Removed):\n", res_base["filtered"])

# Exercise 9: Evaluation Across 3 Domain Statements
test_sentences = [
    "Artificial intelligence systems are rapidly transforming healthcare, finance, and autonomous transportation.",
    "Machine learning models require clean, normalized, and preprocessed datasets to make accurate predictions.",
    "Natural language processing enables computers to interpret human language through tokenization and semantic analysis."
]

print("\n=== EXERCISE 9: EVALUATION ACROSS THREE DOMAIN STATEMENTS ===")
for idx, s in enumerate(test_sentences, 1):
    res = preprocess_text(s)
    print(f"\nSentence {idx}: {s}")
    print(f"Filtered Keywords: {res['filtered']}")
```

```python
from nltk.stem import WordNetLemmatizer

# Exercise 10: Create and load local text file
file_content = """Artificial intelligence algorithms analyze massive volumes of unstructured text data.
Researchers are building complex neural networks to understand natural human languages.
Many developers utilize machine learning techniques because these systems perform better on diverse tasks."""

filename = "sample_ai_text.txt"
with open(filename, "w", encoding="utf-8") as f:
    f.write(file_content)

with open(filename, "r", encoding="utf-8") as f:
    loaded_text = f.read()

print(f"=== EXERCISE 10: TEXT READ FROM {filename} ===")
print(loaded_text)

# Exercise 11: Preprocessing the file contents
file_processed = preprocess_text(loaded_text)
print("\n=== EXERCISE 11: EXTRACTED FILE TOKENS (FILTERED) ===")
print(file_processed["filtered"])

# Exercise 12: Baseline Lemmatizer Demonstration
lemmatizer = WordNetLemmatizer()
sample_words = ['cats', 'cacti', 'geese', 'rocks', 'python', 'better']

print("\n=== EXERCISE 12: BASELINE LEMMATIZER EVALUATION ===")
for w in sample_words:
    lemma = lemmatizer.lemmatize(w)
    print(f"Word: {w:<8} -> Lemma: {lemma}")

# Exercise 13: Lemmatizing the preprocessed file tokens
lemmatized_file_tokens = [lemmatizer.lemmatize(token) for token in file_processed["filtered"]]
print("\n=== EXERCISE 13: LEMMATIZED FILE TOKENS ===")
print(lemmatized_file_tokens)
```

```python
from collections import defaultdict
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords

# Exercise 14: Sentence Segmentation Demo
seg_demo_text = "Hello! This is an example of sentence segmentation. It demonstrates how sentence segmentation works using NLTK. Hope you find it helpful."
sentences_demo = sent_tokenize(seg_demo_text)

print("=== EXERCISE 14: SENTENCE SEGMENTATION ===")
for i, s in enumerate(sentences_demo, 1):
    print(f"[{i}] {s}")

# Exercise 15: Extractive Text Summarizer Implementation
corpus_text = """Artificial Intelligence has revolutionized how modern society interacts with digital technology and automated systems.
Natural Language Processing is a specialized subfield of computer science dedicated to making human speech and text understandable to machines.
Tokenization breaks unstructured text into individual components such as words and punctuation marks.
Stop word removal filters out common non-informative words like articles and prepositions to highlight core terms.
Lemmatization transforms inflected word variants into their canonical lexical roots.
Feature extraction converts clean text tokens into numerical vectors that deep learning algorithms can process.
Automated summarization algorithms calculate the statistical importance of sentences based on keyword frequencies to extract key insights.
Through these integrated preprocessing techniques, artificial intelligence models achieve high accuracy across translation, classification, and sentiment tasks."""

def generate_text_summary(raw_text, top_n=2, max_sentence_len=30):
    # Step A: Sentence tokenization
    sentences = sent_tokenize(raw_text)
    
    # Step B & C: Stop-word filtering & word frequency calculation
    stop_words = set(stopwords.words('english'))
    word_frequencies = defaultdict(int)
    
    for word in word_tokenize(raw_text.lower()):
        if word.isalnum() and word not in stop_words:
            word_frequencies[word] += 1
            
    # Step D: Score sentences based on word frequencies (< 30 words)
    sentence_scores = {}
    for sent in sentences:
        words = word_tokenize(sent.lower())
        if len(words) < max_sentence_len:
            score = sum(word_frequencies[w] for w in words if w in word_frequencies)
            sentence_scores[sent] = score
            
    # Step E: Rank sentences and select the top N candidates
    ranked = sorted(sentence_scores.items(), key=lambda x: x[1], reverse=True)
    selected_sentences = [sent for sent, score in ranked[:top_n]]
    
    # Step F: Assemble summary in natural document order
    final_summary = " ".join([s for s in sentences if s in selected_sentences])
    return final_summary, ranked

summary_output, sentence_rankings = generate_text_summary(corpus_text, top_n=2)

print("\n=== EXERCISE 15: EXTRACTED TEXT SUMMARY (TOP 2 SENTENCES) ===")
print(summary_output)

print("\n=== RANKED SENTENCE IMPORTANCE SCORES ===")
for s, score in sentence_rankings:
    print(f"Score: {score:<3} | Sentence: {s[:65]}...")
```
